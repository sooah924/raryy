```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate median_house_value per population
df['value_per_person'] = df['median_house_value'] / df['population']

plt.figure(figsize=(12, 10))
sns.scatterplot(
    data=df,
    x='longitude',
    y='latitude',
    hue='value_per_person',
    size='value_per_person',
    sizes=(20, 600), # Adjust point sizes for better visualization
    palette='viridis',
    alpha=0.6,
    edgecolor='w',
    linewidth=0.5
)

plt.title('California Housing: Median House Value per Population by Location', fontsize=16)
plt.xlabel('Longitude', fontsize=12)
plt.ylabel('Latitude', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(title='Value per Person', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. 파일 불러오기
path = '/content/sample_data/california_housing_train.csv'
df = pd.read_csv(path)

# 2. 산점도를 통한 지도 데이터 시각화
plt.figure(figsize=(10, 7))

# scatter plot 그리기
# x: 경도(longitude), y: 위도(latitude)
# c: 집값(median_house_value) - 색상 구분
# s: 인구수(population) - 원 크기 구분 (보기 좋게 100으로 나눔)
scatter = plt.scatter(
    x=df['longitude'],
    y=df['latitude'],
    c=df['median_house_value'],
    s=df['population'] / 100,  # 원의 크기를 인구수에 비례하게 설정
    cmap='jet',                # 파란색(저렴) -> 빨간색(비쌈) Color map
    alpha=0.4                  # 점들의 겹침을 보기 위한 투명도 설정
)

# 컬러바 추가 (집값 범위 안내)
cbar = plt.colorbar(scatter)
cbar.set_label('Median House Value ($)', fontsize=12)

# 축 라벨 및 제목 설정
plt.xlabel('Longitude (경도)', fontsize=12)
plt.ylabel('Latitude (위도)', fontsize=12)
plt.title('California Housing Prices & Population Map', fontsize=14)

# 그리드 추가
plt.grid(True, linestyle='--', alpha=0.5)

# 출력
plt.show()
```

목요일과 금요일의 tip 데이터분포를 확인
평균(Mean), 중앙값(Median), 최빈값(Mode)
왜곡(Skeness)
상자박스(Box plot), 히스토그램(Histogram)

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter data for Thursday and Friday
df_filtered = tips[tips['day'].isin(['Thur', 'Fri'])]
# Remove unused categories to match the filtered data and suppress warning
df_filtered['day'] = df_filtered['day'].cat.remove_unused_categories()

# Create a box plot to visualize the tip distribution for 'Thur' and 'Fri'
plt.figure(figsize=(8, 6))
sns.boxplot(x='day', y='tip', hue='day', data=df_filtered, palette=['skyblue', 'lightcoral'], legend=False)
plt.title('Tip Distribution for Thursday and Friday')
plt.xlabel('Day of the Week')
plt.ylabel('Tip Amount')
sns.despine(offset=10, trim=True)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Filter data for Thursday and Friday
df_filtered = tips[tips['day'].isin(['Thur', 'Fri'])]
# Remove unused categories to match the filtered data and suppress warning
df_filtered['day'] = df_filtered['day'].cat.remove_unused_categories()

plt.figure(figsize=(10, 6))
sns.histplot(data=df_filtered, x='tip', hue='day', kde=True, palette=['skyblue', 'lightcoral'])
plt.title('Tip Distribution for Thursday and Friday (Histogram)')
plt.xlabel('Tip Amount')
plt.ylabel('Frequency')
sns.despine(offset=10, trim=True)
plt.show()
```

```python
result_agg = tips.groupby('day', observed=False)['tip'].agg(
    mean='mean',
    median='median',
    mode=lambda x: x.mode()[0] if not x.mode().empty else None
)

result_agg
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 평균, 중앙값, 최빈값을 시각화합니다.
fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)
fig.suptitle('요일별 Tip의 평균, 중앙값, 최빈값 비교', fontsize=16)

# 평균 Tip 시각화
sns.barplot(x=result_agg.index, y=result_agg['mean'], ax=axes[0], palette='viridis')
axes[0].set_title('요일별 평균 Tip')
axes[0].set_xlabel('요일')
axes[0].set_ylabel('평균 Tip 금액')

# 중앙값 Tip 시각화
sns.barplot(x=result_agg.index, y=result_agg['median'], ax=axes[1], palette='plasma')
axes[1].set_title('요일별 중앙값 Tip')
axes[1].set_xlabel('요일')
axes[1].set_ylabel('중앙값 Tip 금액')

# 최빈값 Tip 시각화
sns.barplot(x=result_agg.index, y=result_agg['mode'], ax=axes[2], palette='cividis')
axes[2].set_title('요일별 최빈값 Tip')
axes[2].set_xlabel('요일')
axes[2].set_ylabel('최빈값 Tip 금액')

plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust layout to prevent suptitle overlap
plt.show()
```
