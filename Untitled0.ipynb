```python
import gradio as gr

def greet(name, intensity):
    return "Hello, " + name + "!" * int(intensity)
```

```python
greet('초',10)
```

```python
demo = gr.Interface(
    fn=greet,
    inputs=["text", "slider"],
    outputs=["text"],
    api_name="predict"
)
```
